```python
# ==============================================================================
# LANGKAH 1: PENGATURAN AWAL & DIREKTORI LOKAL
# ==============================================================================
print("="*80)
print("LANGKAH 1: PENGATURAN AWAL & DIREKTORI LOKAL")
print("="*80)

import os
import shutil
import glob
import random
import yaml
import cv2
import numpy as np
import re
import zipfile
import json
import time
from tqdm.notebook import tqdm
# Hapus impor spesifik Google Colab
# from google.colab import drive, files
# from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

# @markdown ### 📂 Tentukan Nama Proyek Anda
PROJECT_NAME = "WhiteFlies_5"  #@param {type:"string"}

# Atur DRIVE_PROJECT_PATH ke direktori kerja lokal Anda di server.
# Ini harus sesuai dengan direktori tempat Anda menyimpan notebook dan data proyek.
# Misalnya, jika Anda mengatur root_dir JupyterLab ke /home/killerking/jupyterlab_project/notebooks,
# maka Anda bisa menggunakan path relatif dari sana atau path absolut.
# Sesuaikan '/home/<USERNAME_SERVER_ANDA>/jupyterlab_project/notebooks' dengan path yang benar di server Anda.
# Contoh:
BASE_NOTEBOOKS_DIR = "/home/killerking/jupyterlab_project/notebooks" # Ganti <USERNAME_SERVER_ANDA>
DRIVE_PROJECT_PATH = os.path.join(BASE_NOTEBOOKS_DIR, PROJECT_NAME)

os.makedirs(DRIVE_PROJECT_PATH, exist_ok=True)
print(f"✅ Folder proyek lokal diatur ke: {DRIVE_PROJECT_PATH}")

# Pindah ke direktori kerja utama proyek.
# Menggunakan os.chdir() lebih portabel daripada %cd di luar lingkungan IPython/Jupyter.
os.chdir(DRIVE_PROJECT_PATH)
print(f"Direktori kerja saat ini: {os.getcwd()}")

# Impor Roboflow setelah instalasi
from roboflow import Roboflow

# ==============================================================================
# LANGKAH 2: PERSIAPAN DATASET (MENGgunakan ANOTASI ASLI ROBOFLOW)
# ==============================================================================
print("\n" + "="*80)
print("LANGKAH 2: PERSIAPAN DATASET (MENGgunakan ANOTASI ASLI ROBOFLOW)")
print("="*80)

# --- Definisi Nama dan Path ---
DATASET_DIR = f"{PROJECT_NAME}_YOLO_Original_Annotations"
YAML_PATH = os.path.join(DATASET_DIR, "data.yaml")

# --- Fungsi Validasi Visual ---
def display_samples_with_boxes(image_folder, label_folder, class_names, title, num_samples=5):
    print(f"\n--- {title} ---")
    if not os.path.exists(image_folder):
        print(f"❌ Peringatan: Direktori gambar tidak ditemukan di '{image_folder}'.")
        return
    image_files = glob.glob(os.path.join(image_folder, '*.*'))
    if not image_files:
        print(f"Tidak ada gambar yang ditemukan di '{image_folder}'.")
        return

    num_to_sample = min(num_samples, len(image_files))
    if num_to_sample == 0: return

    random_samples = random.sample(image_files, num_to_sample)
    plt.figure(figsize=(25, 6))
    plt.suptitle(title, fontsize=16)

    for i, img_path in enumerate(random_samples):
        plt.subplot(1, num_to_sample, i + 1)
        img = cv2.imread(img_path)
        label_filename = os.path.splitext(os.path.basename(img_path))[0] + ".txt"
        label_path = os.path.join(label_folder, label_filename)

        if os.path.exists(label_path):
            with open(label_path, 'r') as f:
                lines = f.readlines()
            for line in lines:
                parts = line.strip().split()
                if len(parts) == 5:
                    class_id, x_c, y_c, w, h = map(float, parts)
                    label_text = class_names[int(class_id)]
                    img_h, img_w = img.shape[:2]
                    x1, y1 = int((x_c - w/2) * img_w), int((y_c - h/2) * img_h)
                    x2, y2 = int((x_c + w/2) * img_w), int((y_c + h/2) * img_h)
                    cv2.rectangle(img, (x1, y1), (x2, y2), (36, 255, 12), 2)
                    cv2.putText(img, label_text, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (36, 255, 12), 2)

        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img_rgb) # Ganti cv2_imshow dengan plt.imshow
        plt.title(os.path.basename(img_path), fontsize=8)
        plt.axis('off')
    plt.show()

# --- LOGIKA UTAMA PERSIAPAN DATA ---
if os.path.exists(YAML_PATH):
    print(f"✅ Folder dataset '{DATASET_DIR}' sudah ada. Melewati persiapan data.")
else:
    # 1. Unduh dataset dari Roboflow
    print("Mengunduh dataset dari Roboflow...")
    api_key_input=os.environ.get("ROBOFLOW_API_KEY", "")  # isi lewat env var ROBOFLOW_API_KEY
    workspace_input="naufalskripsi" #@param {type:"string"}
    project_input="whiteflies-segmentation" #@param {type:"string"}
    project_version_input=10 #@param {type:"number"}
    dataset_version_type="yolov11" #@param {type:"string"}

    rf = Roboflow(api_key=api_key_input)
    project = rf.workspace(workspace_input).project(project_input)
    version = project.version(project_version_input)

    # Unduh langsung ke direktori tujuan
    if os.path.exists(DATASET_DIR): shutil.rmtree(DATASET_DIR)
    dataset = version.download(dataset_version_type, location=DATASET_DIR)

    YAML_PATH = os.path.join(dataset.location, "data.yaml")
    print(f"✅ Dataset asli berhasil diunduh ke: {dataset.location}")

    # Koreksi path di dalam file data.yaml agar portabel.
    print("\n⚙️ Mengoreksi path dalam file data.yaml agar portabel...")
    with open(YAML_PATH, 'r') as f: data_yaml = yaml.safe_load(f)

    data_yaml['path'] = os.path.abspath(dataset.location)

    with open(YAML_PATH, 'w') as f: yaml.dump(data_yaml, f, sort_keys=False)
    print("✅ Path dalam data.yaml berhasil dikoreksi.")

# --- VALIDASI HASIL PELABELAN ---
if os.path.exists(YAML_PATH):
    print("\n" + "="*80)
    print("VALIDASI DATASET DENGAN ANOTASI ASLI: MENAMPILKAN 5 CONTOH")
    print("="*80)
    with open(YAML_PATH, 'r') as f: data_yaml = yaml.safe_load(f)
    final_class_names = data_yaml['names']

    # Cari nama folder valid/val
    val_dir_name = 'valid'
    if not os.path.exists(os.path.join(data_yaml['path'], 'valid')):
        if os.path.exists(os.path.join(data_yaml['path'], 'val')):
            val_dir_name = 'val'

    paths = {
        "train_images": os.path.join(data_yaml['path'], "train", "images"),
        "train_labels": os.path.join(data_yaml['path'], "train", "labels"),
        "valid_images": os.path.join(data_yaml['path'], val_dir_name, "images"),
        "valid_labels": os.path.join(data_yaml['path'], val_dir_name, "labels"),
        "test_images": os.path.join(data_yaml['path'], "test", "images"),
        "test_labels": os.path.join(data_yaml['path'], "test", "labels")
    }

    display_samples_with_boxes(paths['train_images'], paths['train_labels'], final_class_names, "Contoh dari Set 'Train'")
    display_samples_with_boxes(paths['valid_images'], paths['valid_labels'], final_class_names, "Contoh dari Set 'Valid' (Validasi)")
    display_samples_with_boxes(paths['test_images'], paths['test_labels'], final_class_names, "Contoh dari Set 'Test'")

# ==============================================================================
# LANGKAH 3: PELATIHAN MODEL YOLO
# ==============================================================================
print("\n" + "="*80)
print("LANGKAH 3: PELATIHAN MODEL YOLO (MENGgunakan ANOTASI ASLI)")
print("="*80)

from ultralytics import YOLO

# @markdown ### 🔄 Lanjutkan Pelatihan (Opsional)
RESUME_TRAINING = False  #@param {type:"boolean"}

if os.path.exists(YAML_PATH):
    model = YOLO("yolo11n.pt")

    results = model.train(
        data=YAML_PATH, # Menggunakan path absolut ke YAML
        epochs=150,
        patience=50,
        imgsz=640,
        optimizer='AdamW',
        project=DRIVE_PROJECT_PATH, # Menggunakan path proyek lokal
        name=f'{PROJECT_NAME}_YOLOv11n_original_annot_results',
        exist_ok=True,
        resume=RESUME_TRAINING
    )

    print(f"\n🎉 Pelatihan selesai. Hasil disimpan di: {results.save_dir}")
else:
    print("❌ Gagal memulai pelatihan. Pastikan Langkah 2 berhasil dijalankan.")

# ==============================================================================
# LANGKAH 4: UJI COBA INFERENSI
# ==============================================================================
print("\n" + "="*80)
print("LANGKAH 4: UJI COBA INFERENSI PADA GAMBAR TES")
print("="*80)

try:
    # Cari folder hasil terbaru
    run_folder_pattern = f'{PROJECT_NAME}_YOLOv11n_original_annot_results*'
    list_of_runs = glob.glob(os.path.join(DRIVE_PROJECT_PATH, run_folder_pattern))
    if not list_of_runs:
            raise FileNotFoundError(f"Tidak ada folder hasil pelatihan yang ditemukan dengan pola: {run_folder_pattern}")
    latest_run_dir = max(list_of_runs, key=os.path.getctime)
    model_best_path = os.path.join(latest_run_dir, 'weights/best.pt')

    if not os.path.exists(model_best_path):
        raise FileNotFoundError(f"File model 'best.pt' tidak ditemukan di {model_best_path}")

    model_inference = YOLO(model_best_path)
    print(f"✅ Model terbaik dimuat dari: {model_best_path}")

    if 'paths' in locals() and 'test_images' in paths and os.path.exists(paths.get('test_images', '')):
        test_images_path = paths['test_images']
        all_test_images = glob.glob(os.path.join(test_images_path, '*.*'))

        if all_test_images:
            random_image_path = random.choice(all_test_images)
            print(f"\n🔎 Melakukan inferensi pada gambar acak: {os.path.basename(random_image_path)}")

            inference_results = model_inference.predict(source=random_image_path, conf=0.25)

            result_image_with_boxes = inference_results[0].plot()
            print("\n🖼️ Hasil Deteksi:")
            # Ganti cv2_imshow dengan matplotlib
            plt.figure(figsize=(10, 10))
            plt.imshow(cv2.cvtColor(result_image_with_boxes, cv2.COLOR_BGR2RGB))
            plt.axis('off')
            plt.show()
        else:
            print(f"❌ Tidak ada gambar yang ditemukan di folder tes: {test_images_path}")
    else:
        print("❌ Tidak dapat menemukan path folder gambar tes. Pastikan Langkah 2 berhasil.")
except Exception as e:
    print(f"❌ Error saat inferensi: {e}. Pastikan Langkah 3 telah dijalankan.")

# ==============================================================================
# LANGKAH 5: EKSPOR & UNDUH ARTEFAK
# ==============================================================================
print("\n" + "="*80)
print("LANGKAH 5: EKSPOR ARTEFAK (PYTORCH)") # Ubah judul
print("="*80)

try:
    # Cari folder hasil terbaru
    run_folder_pattern = f'{PROJECT_NAME}_YOLOv11n_original_annot_results*'
    list_of_runs = glob.glob(os.path.join(DRIVE_PROJECT_PATH, run_folder_pattern))
    if not list_of_runs:
            raise FileNotFoundError(f"Tidak ada folder hasil pelatihan yang ditemukan dengan pola: {run_folder_pattern}")
    latest_run_dir = max(list_of_runs, key=os.path.getctime)
    model_best_path = os.path.join(latest_run_dir, 'weights/best.pt')

    if not os.path.exists(model_best_path):
        raise FileNotFoundError(f"File model 'best.pt' tidak ditemukan di {model_best_path}")

    # Siapkan folder ekspor
    export_base_path = os.path.join(DRIVE_PROJECT_PATH, f"{PROJECT_NAME}_Original_Annot_Artifacts")
    os.makedirs(export_base_path, exist_ok=True)

    # Salin bobot PyTorch terbaik
    print(f"    -> Menyalin model terbaik: {os.path.basename(model_best_path)}")
    shutil.copy(model_best_path, os.path.join(export_base_path, "best.pt"))

    # Salin file YAML konfigurasi dataset
    if os.path.exists(YAML_PATH):
        print(f"    -> Menyalin file konfigurasi: {os.path.basename(YAML_PATH)}")
        shutil.copy(YAML_PATH, os.path.join(export_base_path, "data.yaml"))

        with open(YAML_PATH, 'r') as f:
            class_names_from_yaml = yaml.safe_load(f)['names']
        labels_dict = {"labels": class_names_from_yaml}
        with open(os.path.join(export_base_path, "labels.json"), 'w') as f_json:
            json.dump(labels_dict, f_json, indent=4)
        print("    -> Berhasil membuat labels.json.")
    else:
        print("    -> ❌ Peringatan: File 'data.yaml' tidak ditemukan untuk membuat 'labels.json'.")


    # Buat file zip
    archive_name = f"{PROJECT_NAME}_Original_Annot_Artifacts"
    final_zip_path = os.path.join(DRIVE_PROJECT_PATH, f"{archive_name}.zip")
    shutil.make_archive(os.path.join(DRIVE_PROJECT_PATH, archive_name), 'zip', export_base_path)
    print(f"\n✅ File '{archive_name}.zip' berhasil dibuat di server Anda pada: {final_zip_path}")

    print("\n--- Artefak Pelatihan Tersedia di Server Anda ---")
    print("Anda dapat mengunduhnya menggunakan SCP, SFTP, atau alat transfer file lainnya.")
    # Hapus baris files.download yang spesifik Colab
    # files.download(os.path.join(DRIVE_PROJECT_PATH, f"{archive_name}.zip"))

except Exception as e:
    print(f"❌ Terjadi kesalahan saat ekspor artefak: {e}")
```
